```python
import chromadb
from pathlib import Path
from sentence_transformers import SentenceTransformer

# 1. Setup paths
PROJECT_ROOT = Path(
    r"c:\Users\Abin ag\Downloads\Sales-Operations-Intelligence-Copilot"
)
CHROMA_DIR = PROJECT_ROOT / "data" / "vector_db"

# 2. Connect to database and access collection
client = chromadb.PersistentClient(path=str(CHROMA_DIR))
collection = client.get_collection(name="company_knowledge")

# 3. Load embedding model
print("Loading embedding model...")
embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# 4. Test question
query = "What products and services does the company provide?"

# 5. Generate query embedding
query_embedding = embedding_model.encode(query).tolist()

# 6. Retrieve relevant chunks
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

# 7. Print results
print("\nQUESTION:", query)
print("\nRETRIEVED DOCUMENTS:\n")

if results and results["documents"] and results["documents"][0]:
    for i, document in enumerate(results["documents"][0], 1):
        # Extract metadata if available
        source = results["metadatas"][0][i-1].get("source", "Unknown") if results["metadatas"] else "Unknown"
        
        print(f"RESULT {i} (Source: {source})")
        print(document)
        print("-" * 60)
else:
    print("No relevant documents found or the collection is currently empty.")
```

```python
from src.rag.retriever import search_documents
from src.llm.ollama_client import generate_response


def ask_company_assistant(question):

    # 1. Retrieve relevant company documents
    retrieved_docs = search_documents(
        query=question,
        top_k=3
    )

    if not retrieved_docs:
        return {
            "question": question,
            "answer": "I couldn't find relevant information in the company documents.",
            "citations": []
        }

    # 2. Build context for the LLM
    context_parts = []

    for i, doc in enumerate(retrieved_docs, start=1):

        context_parts.append(
            f"""
[Source {i}]
Filename: {doc['source']}
Chunk ID: {doc['chunk_id']}
Content:
{doc['content']}
"""
        )

    context = "\n".join(context_parts)

    # 3. Grounded prompt
    prompt = f"""
You are a professional company knowledge assistant.

Answer the user's question using ONLY the company context below.

Rules:
1. Do not invent company information.
2. If the answer is unavailable, clearly say so.
3. Keep the answer professional and concise.
4. Cite supporting sources using their exact labels,
   such as [Source 1] or [Source 2].
5. Do not cite a source unless it supports the claim.

COMPANY CONTEXT:
{context}

USER QUESTION:
{question}

ANSWER:
"""

    # 4. Generate answer
    answer = generate_response(prompt)

    # 5. Prepare detailed citations
    citations = []

    for i, doc in enumerate(retrieved_docs, start=1):

        citations.append({
            "citation": f"Source {i}",
            "filename": doc["source"],
            "chunk_id": doc["chunk_id"],
            "distance": doc["distance"],
            "supporting_text": doc["content"]
        })

    return {
        "question": question,
        "answer": answer,
        "citations": citations
    }


if __name__ == "__main__":

    question = "What services does the company offer?"

    result = ask_company_assistant(question)

    print("\n" + "=" * 65)
    print("COMPANY KNOWLEDGE ASSISTANT")
    print("=" * 65)

    print("\nQUESTION:")
    print(result["question"])

    print("\nAI ANSWER:")
    print(result["answer"])

    print("\nDETAILED SOURCE CITATIONS:")

    for citation in result["citations"]:

        print("\n" + "-" * 50)

        print("Citation:", citation["citation"])
        print("Filename:", citation["filename"])
        print("Chunk ID:", citation["chunk_id"])
        print("Distance:", citation["distance"])

        print("\nSupporting Text:")
        print(citation["supporting_text"])

    print("\n" + "=" * 65)
```

```python
import sys
import gc
from pathlib import Path

# 1. Project path configurations
PROJECT_ROOT = Path(r"c:\Users\Abin ag\Downloads\Sales-Operations-Intelligence-Copilot")
VENV_PACKAGES = PROJECT_ROOT / ".venv" / "Lib" / "site-packages"

if str(VENV_PACKAGES) not in sys.path:
    sys.path.insert(0, str(VENV_PACKAGES))
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

# 2. Extract context first
from src.rag.retriever import search_documents

question = "What services does the company offer?"
print("Extracting records from vector store...")
retrieved_docs = search_documents(query=question, top_k=2)

# 3. Clean local memory footprints
gc.collect()

# 4. Compile context elements
context_parts = []
for i, doc in enumerate(retrieved_docs, start=1):
    content = doc.get("content") or doc.get("document") or str(doc) if isinstance(doc, dict) else getattr(doc, "page_content", str(doc))
    context_parts.append(f"[Source {i}]: {content}")
context = "\n".join(context_parts)

prompt = f"Context:\n{context}\n\nQuestion: {question}\nAnswer using only context:"

# 5. Direct Cloud API generation call
print("Requesting cloud inference for Llama 3.2 3B...")
import ollama

try:
    # Adding :cloud offloads the processing overhead from your laptop's RAM
    response = ollama.generate(
        model="llama3.2:cloud",
        prompt=prompt,
        options={
            "temperature": 0.0
        }
    )
    print("\n" + "=" * 65)
    print("AI ASSISTANT ANSWER:")
    print("=" * 65)
    print(response['response'])
except Exception as e:
    print(f"\nExecution hit an error: {e}")
```
